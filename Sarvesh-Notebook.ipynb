```r
library(ISLR)
library(ggplot2)

hours <- c(1,2,3,3,4,5,6,7,8,9)
score <- c(63,56,66,67,70,72,86,81,86,90)

exam <- data.frame(hours, score)

cor(exam$hours, exam$score)

ggplot(exam, aes(x = hours, y= score)) +
    geom_point(size = 3) +
    labs(
        title = "Hours Studied vs. Exam Score", 
        x = "Hours Studied", 
        y = "Exam Score"
    ) +
    theme_minimal()


model <- lm(score ~ hours, data = exam)

ggplot(exam, aes(x = hours, y= score)) +
    geom_point(size = 3) +
    geom-smooth(method = "lm"
    labs(
        title = "Hours Studied vs. Exam Score", 
        x = "Hours Studied", 
        y = "Exam Score"
    ) +
    theme_minimal()

predicted <- predict(model)

residuals <- exam$score - predicted

rss <- sum(residuals^2)

cat("sum of residual squares: ", rss)
```

```r
library(ISLR)
library(ggplot2)

# Create the dataset
df <- data.frame(
  Student = 1:10,
  Hours_Studied = c(1, 2, 3, 3, 4, 5, 6, 7, 8, 9),
  Exam_Score = c(63, 56, 66, 67, 70, 72, 86, 81, 86, 90)
)

# Display the data frame
print(df)

# Calculate the Pearson correlation coefficient
correlation <- cor(df$Hours_Studied, df$Exam_Score)
cat("Pearson Correlation Coefficient (r):", correlation, "\n")

# Base R scatter plot to visualize the raw data
plot(df$Hours_Studied, df$Exam_Score, 
     main = "Pre-Regression Check: Hours Studied vs. Exam Score",
     xlab = "Hours Studied (x)", 
     ylab = "Exam Score (y)", 
     pch = 16, 
     col = "darkblue",
     cex = 1.3)

# Fit the linear model
model <- lm(Exam_Score ~ Hours_Studied, data = df)

# Extract specific coefficients
coefficients <- coef(model)
intercept <- coefficients[1]
slope <- coefficients[2]

# Display results neatly
cat("--- Model Coefficients ---\n")
cat("Intercept (b0):", round(intercept, 4), "\n")
cat("Slope (b1):    ", round(slope, 4), "\n\n")

# Optional: View the full summary for p-values and R-squared
summary(model)

# Plot the data points again
plot(df$Hours_Studied, df$Exam_Score, 
     main = "Linear Regression Model Fit",
     xlab = "Hours Studied (x)", 
     ylab = "Exam Score (y)", 
     pch = 16, 
     col = "darkblue",
     cex = 1.3)

# Add the calculated linear regression line using abline()
abline(model, col = "red", lwd = 2.5)

# Add a subtle grid for better readability
grid(nx = NULL, ny = NULL, col = "lightgray", lty = "dotted")
```
